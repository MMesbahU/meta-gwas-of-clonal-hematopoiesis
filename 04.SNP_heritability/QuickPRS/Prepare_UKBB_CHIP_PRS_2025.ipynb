```r

### CHIP PRS #####

## PRS with AFR Reference Panel
# List PRS files
prs_path <- "afr/"

list_prs_score_afr <- list.files(prs_path, pattern = "\\.profile$", full.names = TRUE)

names_prs_score_afr_tmp <- gsub(".*has(.*)\\.profile$", "\\1", list_prs_score_afr)

head(names_prs_score_afr_tmp)
table(table(names_prs_score_afr_tmp))
names_prs_score_afr <- paste(stringr::str_split_fixed(string = names_prs_score_afr_tmp, pattern = "[.]",n = 3)[,1], 
      stringr::str_split_fixed(string = names_prs_score_afr_tmp, pattern = "[.]",n = 3)[,2], sep="_")


head(list_prs_score_afr)
head(names_prs_score_afr)
table(table(names_prs_score_afr))
```

```r


list_prs_score.afr <- system("ls afr/*.profile | awk '{print $NF}'", 
                             intern = TRUE)

names_prs_score.afr <- system("ls /medpop/esp2/mesbah/projects/Meta_GWAS/MetaGWAS_N900k/sumHer/prs/scores/afr/*.profile | awk '{print $NF}' | cut -d '.' -f7,8  | sed 's:has::g' | tr '.' '_' ", 
                              intern = TRUE)
## Read PRS files
d_afr <- fread(list_prs_score.afr[1], header=T, select = c("ID2", "Profile_1"))

str(d_afr)

names(d_afr)

summary(d_afr)

names(d_afr) <- c("ID2", names_prs_score.afr[1])

str(d_afr)

dat_afr <- d_afr; rm(d_afr)

for(i in 2:length(list_prs_score.afr)){
  print(i)
  
  print(list_prs_score.afr[i])
  
  print(names_prs_score.afr[i])
  
  d <- fread(list_prs_score.afr[i], header=T, select = c("ID2", "Profile_1"))
  
  names(d) <- c("ID2", names_prs_score.afr[i])
  
  dat_afr <- merge(dat_afr, d, by="ID2")
  
  rm(d)
  
}

str(dat_afr)

head(dat_afr)
```
