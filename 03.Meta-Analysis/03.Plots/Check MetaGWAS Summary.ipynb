```r
# Define the sizes and thresholds
size_n <- c(300000, 400000, 500000, 600000)
thresholds <- list(
  eaf_1 = c(0.001, Inf),
  eaf_2 = c(0.01, 0.99),
  eaf_3 = c(0.005, 0.01),
  eaf_4 = c(0.001, 0.005)
)

# Initialize an empty list to store results
results <- list()

# Loop through sample sizes and thresholds
for (size in size_n) {
  for (thresh_name in names(thresholds)) {
    eaf_range <- thresholds[[thresh_name]]
    
    # Filter the data
    valid_indices <- dat_2$eaf >= eaf_range[1] & dat_2$eaf < eaf_range[2] & dat_2$n_samples >= size
    valid_p <- dat$`p-value`[valid_indices]
    
    # Calculate GWAS Lambda
    lambda <- median(qchisq(1 - valid_p, 1), na.rm = TRUE) / qchisq(0.5, 1)
    
    # Store the result
    results[[paste0("size_", size, "_", thresh_name)]] <- lambda
  }
}

# Convert results to a data frame for better readability
results_df <- data.frame(
  scenario = names(results),
  lambda = unlist(results)
)

# Print the results
print(results_df)
```

```r
# library(dplyr)

# Assuming your data is in a data frame called d
# Create a new column for chromosome and position from the 'varid' column
d <- d %>%
  mutate(
    chr = as.numeric(sub("chr(\\d+):.*", "\\1", rs_number)),
    pos = as.numeric(sub("chr\\d+:(\\d+):.*", "\\1", rs_number))
  )

# Filter for significant variants and then group by regions within ±500kb
result <- d %>%
  filter(
    `p-value` < 5e-8,
    n_studies >= 2,
    (eaf < 0.01 | eaf > 0.99)
  ) %>%
  arrange(chr, pos) %>% # Arrange by chromosome and position
  group_by(group = cumsum(c(0, diff(pos)) > 500000 | c(0, diff(chr)) != 0)) %>% # Group by regions within ±500kb
  summarise(
    variants = paste(rs_number, collapse = ", "),
    .groups = "drop"
  )

# Display the grouped result
print(result)
```
