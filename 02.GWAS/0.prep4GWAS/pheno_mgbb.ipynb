## Mass General Brigham Biobank

```r
library(data.table)
```

```r
## MGBB CH

pheno.mgbb_gwas <- fread("/medpop/esp2/mesbah/projects/Meta_GWAS/rerun/mgbb53k/mgbb53k.imp_new.noRel_sk.21Jul2022.tsv")
```

```r
## CHIP variants
mgbb_ch <- fread("/medpop/esp2/mesbah/datasets/CHIP/MGBB/MGBB53k_Dropbox/MGBB53k_CHIP_return/MGBB_53k_WES_CHIP_calls_02Mar2023.csv", 
                 header=T)
names(mgbb_ch)

## CHIP calls
mgbb_vars <- fread("/medpop/esp2/mesbah/datasets/CHIP/MGBB/MGBB53k_Dropbox/MGBB53k_CHIP_return/MGBB_53k_WES_CHIP_variants_02Mar2023.csv", 
                 header=T)

names(mgbb_vars)
```

```r
# VAF>=2%
pheno.mgbb_gwas$hasCHvaf02 <- ifelse(pheno.mgbb_gwas$Biobank_Subject_ID %in% 
                                     mgbb_vars$Biobank_Subject_ID[mgbb_vars$VAF>=0.02], 
                                     1, 0)
table(pheno.mgbb_gwas$hasCHvaf02, exclude=NULL)
# VAF>=10%
pheno.mgbb_gwas$hasCHvaf10 <- ifelse(pheno.mgbb_gwas$Biobank_Subject_ID %in% 
                                     mgbb_vars$Biobank_Subject_ID[mgbb_vars$VAF>=0.10], 1, 
                                     ifelse(pheno.mgbb_gwas$hasCHvaf02==0,0,NA))
table(pheno.mgbb_gwas$hasCHvaf10, exclude=NULL)
```

```r
## Gene-specific CH
# DNMT3A
pheno.mgbb_gwas$hasDNMT3A <- ifelse(pheno.mgbb_gwas$Biobank_Subject_ID %in% 
                                     mgbb_vars$Biobank_Subject_ID[mgbb_vars$Gene =="DNMT3A"] & 
                                    pheno.mgbb_gwas$hasCHvaf02==1, 1, 
                                     ifelse(pheno.mgbb_gwas$hasCHvaf02==0,0,NA))
table(pheno.mgbb_gwas$hasDNMT3A, exclude=NULL)

# TET2
 pheno.mgbb_gwas$hasTET2 <- ifelse(pheno.mgbb_gwas$Biobank_Subject_ID %in% 
                                     mgbb_vars$Biobank_Subject_ID[mgbb_vars$Gene =="TET2"] & 
                                    pheno.mgbb_gwas$hasCHvaf02==1, 1, 
                                     ifelse(pheno.mgbb_gwas$hasCHvaf02==0,0,NA))
table(pheno.mgbb_gwas$hasTET2, exclude=NULL)

# ASXL1
 pheno.mgbb_gwas$hasASXL1 <- ifelse(pheno.mgbb_gwas$Biobank_Subject_ID %in% 
                                     mgbb_vars$Biobank_Subject_ID[mgbb_vars$Gene =="ASXL1"] & 
                                    pheno.mgbb_gwas$hasCHvaf02==1, 1, 
                                     ifelse(pheno.mgbb_gwas$hasCHvaf02==0,0,NA))
table(pheno.mgbb_gwas$hasASXL1, exclude=NULL)

# SF: c("SF3B1","SRSF2", "U2AF1", "ZRSR2")
pheno.mgbb_gwas$hasSF <- ifelse(pheno.mgbb_gwas$Biobank_Subject_ID %in% 
                                     mgbb_vars$Biobank_Subject_ID[mgbb_vars$Gene %in% 
                                                                  c("SF3B1","SRSF2", "U2AF1", "ZRSR2")] & 
                                    pheno.mgbb_gwas$hasCHvaf02==1, 1, 
                                     ifelse(pheno.mgbb_gwas$hasCHvaf02==0,0,NA))
table(pheno.mgbb_gwas$hasSF, exclude=NULL)

# DDR: c("PPM1D", "TP53")
pheno.mgbb_gwas$hasDDR <- ifelse(pheno.mgbb_gwas$Biobank_Subject_ID %in% 
                                     mgbb_vars$Biobank_Subject_ID[mgbb_vars$Gene %in% 
                                                                  c("PPM1D", "TP53")] & 
                                    pheno.mgbb_gwas$hasCHvaf02==1, 1, 
                                     ifelse(pheno.mgbb_gwas$hasCHvaf02==0,0,NA))
table(pheno.mgbb_gwas$hasDDR, exclude=NULL)
```
