```r
library(data.table)
library(dplyr)

setwd("/medpop/esp2/mesbah/projects/Meta_GWAS/MetaGWAS_N900k/topmed_v10/gwas")
```

## Version 1

```r
## /medpop/esp/mesbah/GWAS_CHIP/topmed_GWAS//topmed2021/topmed_hasCHIP_2020_08_31.tsv.gz
top2020 <- fread("MultiANC/2020_vs_2024/run2020.topmed2021_hasCH.p5e6maf005.tsv", 
                 header=T)

head(top2020)

names(top2020)
```

## Version 2

```r
## /medpop/esp2/mesbah/projects/Meta_GWAS/MetaGWAS_N900k/topmed_v10/gwas/MultiANC/MultiANC.TOPMed74k.chr1_22.hasCHvaf02.regenie.gz
top2024 <- fread("MultiANC/2020_vs_2024/run2024.hasCHvaf02.MultiANC.TOPMed74k.chr1_22.tsv", 
                 header=T)


names(top2024)

top2024$N <- (top2024$Num_Cases + top2024$Num_Controls)

head(top2024)
```

```r
table(top2020$SNPID %in% top2024$SNPID)

table(top2024$SNPID %in% top2020$SNPID)
```

```r
top20_24 <- merge(top2020, 
                  top2024, 
                  by="SNPID", 
                  all=T)

head(top20_24)
```

```r
cor.test( -log10(top20_24$P), -log10(top20_24$Pval) )

plot( -log10(top20_24$P), -log10(top20_24$Pval), log="xy" )
```

```r
cat("GWAS 2020 P<=5e-8\n")

table(top20_24$P<=5e-8, exclude=NULL)

cor.test( -log10(top20_24$P[top20_24$P<=5e-8]), -log10(top20_24$Pval[top20_24$P<=5e-8]) )

plot( -log10(top20_24$P[top20_24$P<=5e-8]), -log10(top20_24$Pval[top20_24$P<=5e-8]), log="xy" )


```

```r
cat("GWAS 2024 P<=5e-8\n")

table(top20_24$Pval<=5e-8, exclude=NULL)

cor.test( -log10(top20_24$P[top20_24$Pval<=5e-8]), -log10(top20_24$Pval[top20_24$Pval<=5e-8]) )

plot( -log10(top20_24$P[top20_24$Pval<=5e-8]), -log10(top20_24$Pval[top20_24$Pval<=5e-8]), log="xy" )
```
