# Annotate 

```r
rm(d)

for(i in 1:length(gwas_files)){
    
    ##
    cat(i,"\n")
    cat(gwas_files[i], "\n")
    ##

    d <- fread(gwas_files[i], header=T)
    
    print(head(d))
    cat("\n")
    cat(nrow(d), "\n")

###
    cat(table(d$SNPID %in% annot$Otherinfo6, exclude=NULL), "\n")

###
    dat <- merge(d, annot, 
                 by.x="SNPID", 
                 by.y="Otherinfo6", 
                 all.x=TRUE)
    cat(nrow(dat), "\n")    
    
###
    outfile_name <- paste0("/medpop/esp2/mesbah/projects/Meta_GWAS/MetaGWAS_N900k/GWAMA/summary/annovar/annovar.", 
                           gsub(pattern = "tsv.gz", 
                           replacement = "csv", 
                           x = gwas_files[i]))
    
    fwrite(x = dat, file = outfile_name, 
           row.names = FALSE, 
           col.names = TRUE, 
           sep=",", 
           quote=TRUE)
    ##
    rm(dat)
    
}
```

## EAF>0.1%

```r
rm(dat, d, outfile_name)
for(i in 1:length(gwas_files)){
    
    ##
    cat(i,"\n")
    cat(gwas_files[i], "\n")
    ##

    d <- fread(gwas_files[i], header=T)
    
    print(head(d))
    cat("\n")
    cat("#SNP in GWAS Summary: ",nrow(d), "\n")

###
    cat("Overall with Annotation: ",table(d$SNPID %in% annot$Otherinfo6, exclude=NULL), "\n")

###
    dat <- merge(d, annot, 
                 by.x="SNPID", 
                 by.y="Otherinfo6", 
                 all.x=TRUE)
    cat("Combined ",nrow(dat), "\n")    
    
    dat <- dat %>% filter(AAF>0.001 & AAF<0.999 & P<5e-7)
    
    cat(nrow(dat), "\n")
###
    outfile_name <- paste0("/medpop/esp2/mesbah/projects/Meta_GWAS/MetaGWAS_N900k/GWAMA/summary/annovar/eaf001/eaf001p5e7.annovar.", 
                           gsub(pattern = "tsv.gz", 
                           replacement = "csv", 
                           x = gwas_files[i]))
    
    fwrite(x = dat, file = outfile_name, 
           row.names = FALSE, 
           col.names = TRUE, 
           sep=",", 
           quote=TRUE)
    ##
    rm(dat, d, outfile_name)
    
}
```

## EAF<=0.1% 

```r
rm(d, dat)

for(i in 1:length(gwas_files)){
    
    ##
    cat(i,"\n")
    cat(gwas_files[i], "\n")
    ##

    d <- fread(gwas_files[i], header=T)
    
    print(head(d))
    cat("\n")
    cat("#SNP in GWAS Summary: ",nrow(d), "\n")

###
    cat("Overall with Annotation: ",table(d$SNPID %in% annot$Otherinfo6, exclude=NULL), "\n")

###
    dat <- merge(d, annot, 
                 by.x="SNPID", 
                 by.y="Otherinfo6", 
                 all.x=TRUE)
    cat("Combined ",nrow(dat), "\n")    
    
    dat <- dat %>% filter(P<5e-7 & (AAF<=0.001 | AAF>=0.999) )
    cat(nrow(dat), "\n")
###
    outfile_name <- paste0("/medpop/esp2/mesbah/projects/Meta_GWAS/MetaGWAS_N900k/GWAMA/summary/annovar/eafBlow001/blow_eaf001p5e7.annovar.", 
                           gsub(pattern = "tsv.gz", 
                           replacement = "csv", 
                           x = gwas_files[i]))
    
    fwrite(x = dat, file = outfile_name, 
           row.names = FALSE, 
           col.names = TRUE, 
           sep=",", 
           quote=TRUE)
    ##
    rm(dat,d, outfile_name)
    
}
```
